```python
import sys
import os
import re
import numpy as np
import tifffile

sys.path.insert(0, '.')
folder_path = 'onesingleGBMcell'

def load_slim_folder(folder):
    files = [f for f in os.listdir(folder) if f.endswith('.tif')]
    
    def get_i_index(filename):
        match = re.search(r'_i(\d+)', filename)
        return int(match.group(1)) if match else 0
    
    files_sorted = sorted(files, key = get_i_index)
    print(f"Found {len(files_sorted)} files")
    print(f"First: {files_sorted[0]}")
    print(f"Last: {files_sorted[-1]}")
    
    frames = [tifffile.imread(os.path.join(folder, f), is_ome = False) for f in files_sorted]
    phi = np.stack(frames, axis = -1).astype(float)
    print(f"Stack shape: {phi.shape} (rows, cols, frames)")
    return phi

phi = load_slim_folder(folder_path)

import importlib.util
spec = importlib.util.spec_from_file_location("ika", "IKA")

from phasor_dps import (detrend, compute_spatial_fft, assign_k_to_windows,
                       compute_dispersion_per_window, compute_probe_omega,
                       compute_phasor_bootstrap, departure_score, 
                       plot_phasor_panels, plot_dispersion_per_window,
                        load_noise_reference, WINDOWS, WINDOW_COLORS,
                        THRESHOLDS, FRAME_RATE)

phi_noise = load_noise_reference(phi=phi)
delta_phi, _ = detrend(phi)
delta_phi_noise, _ = detrend(phi_noise)
k_bins, DELTA_PHI_k, Sk = compute_spatial_fft(delta_phi_noise)
_, DELTA_PHI_noise, Sk_noise = compute_spatial_fft(delta_phi_noise)
window_masks = assign_k_to_windows(k_bins)
dispersion = compute_dispersion_per_window(k_bins, DELTA_PHI_k, window_masks)
n_frames = delta_phi.shape[2]
omega = compute_probe_omega(n_frames)
g_mean, g_std, s_mean, s_std, d_mean, d_std = compute_phasor_bootstrap(DELTA_PHI_k, omega)
fig_phasor = plot_phasor_panels(k_bins, g_mean, s_mean, Sk, Sk_noise,
                               window_masks, d_mean, d_std)
fig_disp = plot_dispersion_per_window(dispersion)

import matplotlib.pyplot as plt
plt.show()
```

```python
from scipy.fft import fft, fftfreq
import numpy as np

def compute_phasor_fixed(DELTA_PHI_k, omega, frame_rate = FRAME_RATE):
    n_k, n_frames = DELTA_PHI_k.shape
    freqs = fftfreq(n_frames, d=1.0/frame_rate)
    omegas_arr = 2 * np.pi * freqs
    omega_idx = np.argmin(np.abs(omegas_arr - omega))
    
    g_arr = np.zeros(n_k)
    s_arr = np.zeros(n_k)
    
    for i in range(n_k):
        ts = DELTA_PHI_k[i]
        std = np.std(ts)
        if std == 0:
            continue
        ts_norm = ts/std
        F = fft(ts_norm)
        
        N = np.abs(F[omega_idx])
        if N == 0:
            continue
        g_arr[i] = np.real(F[omega_idx]) / np.sum(np.abs(F[1:]))
        s_arr[i] = np.imag(F[omega_idx]) / np.sum(np.abs(F[1:]))
        
    return g_arr, s_arr
    
g_fixed, s_fixed = compute_phasor_fixed(DELTA_PHI_k, omega)
print("g range:", g_fixed.min(), g_fixed.max())
print("s range:", s_fixed.min(), s_fixed.max())
print("g values:", g_fixed[:5])
print("s values:", s_fixed[:5])
```

```python
# Sweep probe frequencies to find where phasor signal is strongest
from scipy.fft import fft, fftfreq
import numpy as np
import matplotlib.pyplot as plt

n_frames = DELTA_PHI_k.shape[1]
freqs = fftfreq(n_frames, d=1.0/FRAME_RATE)
omegas_arr = 2 * np.pi * freqs
pos_freqs = freqs[freqs > 0]
pos_omegas = omegas_arr[freqs > 0]

# For first passing k-bin, plot power spectrum
ratio = Sk_normalized / Sk_noise_normalized
passing_idx = np.where(ratio > 1)[0]
first_passing = passing_idx[0]

ts = DELTA_PHI_k[first_passing]
std = np.std(ts)
ts_norm = ts / std
F = fft(ts_norm)
power = np.abs(F[freqs > 0])**2

plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.plot(pos_freqs, power)
plt.xlabel('Frequency (Hz)')
plt.ylabel('Power')
plt.title('Power spectrum of first passing k-bin')
plt.yscale('log')

# Find peak frequency
peak_idx = np.argmax(power)
peak_freq = pos_freqs[peak_idx]
peak_omega = pos_omegas[peak_idx]
print(f"Peak frequency: {peak_freq:.4f} Hz")
print(f"Peak omega: {peak_omega:.4f} rad/s")
print(f"Fundamental frequency: {FRAME_RATE/n_frames:.4f} Hz")

# Recompute phasor at peak frequency
def compute_phasor_at_omega(DELTA_PHI_k, target_omega, frame_rate=FRAME_RATE):
    n_k, n_frames = DELTA_PHI_k.shape
    freqs = fftfreq(n_frames, d=1.0/frame_rate)
    omegas_arr = 2 * np.pi * freqs
    omega_idx = np.argmin(np.abs(omegas_arr - target_omega))
    g_arr = np.zeros(n_k)
    s_arr = np.zeros(n_k)
    for i in range(n_k):
        ts = DELTA_PHI_k[i]
        std = np.std(ts)
        if std == 0:
            continue
        ts_norm = ts / std
        F = fft(ts_norm)
        total = np.sum(np.abs(F[1:]))
        if total == 0:
            continue
        g_arr[i] = np.real(F[omega_idx]) / total
        s_arr[i] = np.imag(F[omega_idx]) / total
    return g_arr, s_arr

g_peak, s_peak = compute_phasor_at_omega(DELTA_PHI_k, peak_omega)
print(f"\nAt peak frequency:")
print(f"g range: {g_peak.min():.4f} to {g_peak.max():.4f}")
print(f"s range: {s_peak.min():.4f} to {s_peak.max():.4f}")

# Plot phasor at peak frequency
plt.subplot(1,2,2)
theta = np.linspace(0, np.pi, 300)
plt.plot(0.5 + 0.5*np.cos(theta), 0.5*np.sin(theta), 'k--', alpha=0.5)
plt.scatter(g_peak[passing_idx], s_peak[passing_idx], 
            c='steelblue', s=30, alpha=0.7)
plt.xlim(-0.1, 1.1)
plt.ylim(-0.1, 0.6)
plt.xlabel('g')
plt.ylabel('s')
plt.title('Phasor at peak frequency')
plt.tight_layout()
plt.show()
```

```python
def compute_phasor_correct(DELTA_PHI_k, target_omega, frame_rate=FRAME_RATE):
    n_k, n_frames = DELTA_PHI_k.shape
    freqs = fftfreq(n_frames, d=1.0/frame_rate)
    omegas_arr = 2 * np.pi * freqs
    omega_idx = np.argmin(np.abs(omegas_arr - target_omega))
    g_arr = np.zeros(n_k)
    s_arr = np.zeros(n_k)
    for i in range(n_k):
        ts = DELTA_PHI_k[i]
        std = np.std(ts)
        if std == 0:
            continue
        ts_norm = ts / std
        F = fft(ts_norm)
        # Normalize by the DC component (zero frequency term)
        # This is the correct FLIM-style normalization
        DC = np.abs(F[0])
        if DC == 0:
            continue
        g_arr[i] = np.real(F[omega_idx]) / DC
        s_arr[i] = np.imag(F[omega_idx]) / DC
    return g_arr, s_arr

g_correct, s_correct = compute_phasor_correct(DELTA_PHI_k, peak_omega)
print(f"g range: {g_correct.min():.4f} to {g_correct.max():.4f}")
print(f"s range: {s_correct.min():.4f} to {s_correct.max():.4f}")

# Plot
import matplotlib.pyplot as plt
ratio = Sk_normalized / Sk_noise_normalized
passing_idx = np.where(ratio > 1)[0]

theta = np.linspace(0, np.pi, 300)
plt.figure(figsize=(6,5))
plt.plot(0.5 + 0.5*np.cos(theta), 0.5*np.sin(theta), 'k--', alpha=0.5, label='Semicircle')
plt.scatter(g_correct[passing_idx], s_correct[passing_idx],
            c='steelblue', s=40, alpha=0.7)
plt.xlim(-0.1, 1.1)
plt.ylim(-0.6, 0.6)
plt.xlabel('g')
plt.ylabel('s')
plt.title('Phasor at peak frequency (corrected normalization)')
plt.axhline(0, color='grey', lw=0.5)
plt.axvline(0, color='grey', lw=0.5)
plt.legend()
plt.show()
```

```python
# Recompute spatial FFT using magnitude before radial averaging
PIXEL_SIZE = 0.1625
FRAME_RATE = 12.5

import numpy as np
from scipy.fft import fft2, fftfreq
from scipy.signal.windows import tukey

def make_tukey_window_2d(shape, alpha = 0.5):
    rows, cols = shape
    wy = tukey(rows, alpha = alpha)
    wx = tukey(cols, alpha = alpha)
    return np.outer(wy, wx)

def compute_spatial_fft_magnitude(delta_phi, pixel_size=PIXEL_SIZE):
    rows, cols, n_frames = delta_phi.shape
    window_2d = make_tukey_window_2d((rows, cols))
    
    kx = fftfreq(cols, d=pixel_size) * 2 * np.pi
    ky = fftfreq(rows, d=pixel_size) * 2 * np.pi
    KX, KY = np.meshgrid(kx, ky)
    K_MAG = np.sqrt(KX**2 + KY**2)
    
    k_max = np.max(K_MAG)
    k_min_nonzero = np.min(K_MAG[K_MAG > 0])
    n_bins = 50
    bin_edges = np.linspace(k_min_nonzero, k_max, n_bins + 1)
    k_bins_new = 0.5 * (bin_edges[:-1] + bin_edges[1:])
    
    # Store magnitude, not complex value
    DELTA_PHI_k_mag = np.zeros((n_bins, n_frames))
    bin_counts = np.zeros(n_bins, dtype=int)
    
    for t in range(n_frames):
        if t % 10 == 0:
            print(f"  Processing frame {t}/{n_frames}")
        frame = delta_phi[:, :, t] * window_2d
        F = fft2(frame)
        F_mag = np.abs(F)  # Take magnitude BEFORE averaging
        for b in range(n_bins):
            mask = (K_MAG >= bin_edges[b]) & (K_MAG < bin_edges[b+1])
            if np.any(mask):
                DELTA_PHI_k_mag[b, t] = np.mean(F_mag[mask])
                bin_counts[b] = np.sum(mask)
    
    valid = bin_counts > 0
    DELTA_PHI_k_mag = DELTA_PHI_k_mag[valid]
    k_bins_new = k_bins_new[valid]
    Sk_new = np.mean(DELTA_PHI_k_mag**2, axis=1)
    
    print(f"Done. k range: {k_bins_new[0]:.2f} to {k_bins_new[-1]:.2f} rad/um")
    return k_bins_new, DELTA_PHI_k_mag, Sk_new

print("Recomputing spatial FFT with magnitude averaging...")
k_bins2, DELTA_PHI_k2, Sk2 = compute_spatial_fft_magnitude(delta_phi)

# Quick check
ts_check = DELTA_PHI_k2[0]
print(f"\nFirst k-bin time series:")
print(f"  mean: {ts_check.mean():.4e}")
print(f"  std:  {ts_check.std():.4e}")
print(f"  std/mean: {ts_check.std()/ts_check.mean():.4f}")
```

```python
# Build phasor from magnitude-based DELTA_PHI_k2
from scipy.fft import fft, fftfreq
import numpy as np
import matplotlib.pyplot as plt

n_frames = DELTA_PHI_k2.shape[1]
freqs = fftfreq(n_frames, d=1.0/FRAME_RATE)
omegas_arr = 2 * np.pi * freqs

# Find probe frequency - use fundamental
omega_probe = 2 * np.pi * (1.0/n_frames) * FRAME_RATE
omega_idx = np.argmin(np.abs(omegas_arr - omega_probe))
print(f"Probe omega: {omega_probe:.4f} rad/s")

# Compute phasor per k-bin
g_arr = np.zeros(len(k_bins2))
s_arr = np.zeros(len(k_bins2))

for i in range(len(k_bins2)):
    ts = DELTA_PHI_k2[i]
    # Detrend: subtract mean
    ts_dt = ts - np.mean(ts)
    F = fft(ts_dt)
    # Normalize by total power excluding DC
    N = np.sum(np.abs(F[1:])**2)
    if N == 0:
        continue
    g_arr[i] = np.real(F[omega_idx]) / N * np.sum(np.abs(F[1:]))
    s_arr[i] = np.imag(F[omega_idx]) / N * np.sum(np.abs(F[1:]))

print(f"g range: {g_arr.min():.4f} to {g_arr.max():.4f}")
print(f"s range: {s_arr.min():.4f} to {s_arr.max():.4f}")

# Compute noise floor from noise reference
g_noise = np.zeros(len(k_bins2))
s_noise = np.zeros(len(k_bins2))

# Recompute Sk2 noise equivalent
_, DELTA_PHI_noise2, Sk_noise2 = compute_spatial_fft_magnitude(delta_phi_noise)
ratio2 = Sk2 / Sk_noise2

# Plot all four phasor panels
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()
thresholds = [None, 1.0, 2.0, 3.0]

# Color by window
from matplotlib.patches import Patch
window_colors = {'A_membrane': 'steelblue',
                 'B_organelle': 'darkorange', 
                 'C_debris': 'mediumseagreen'}
k_colors = np.array(['lightgrey'] * len(k_bins2), dtype=object)
for name, (k_lo, k_hi) in WINDOWS.items():
    mask = (k_bins2 >= k_lo) & (k_bins2 < k_hi)
    k_colors[mask] = window_colors[name]

s_min, s_max = Sk2.min(), Sk2.max()
dot_sizes = 5 + 80 * (Sk2 - s_min) / (s_max - s_min + 1e-10)

theta = np.linspace(0, np.pi, 300)
sc_g = 0.5 + 0.5*np.cos(theta)
sc_s = 0.5*np.sin(theta)

for idx, thresh in enumerate(thresholds):
    ax = axes[idx]
    ax.plot(sc_g, sc_s, 'k--', lw=1, alpha=0.5)
    ax.set_xlim(-0.1, 1.1)
    ax.set_ylim(-0.1, 0.65)
    ax.set_xlabel('g')
    ax.set_ylabel('s')

    if thresh is None:
        keep = np.ones(len(k_bins2), dtype=bool)
        ax.set_title('Plot 0: all k-bins')
    else:
        keep = ratio2 > thresh
        ax.set_title(f'Plot {idx}: S(k) > {thresh:.0f}x noise ({keep.sum()}/{len(k_bins2)} bins)')

    if np.any(keep):
        ax.scatter(g_arr[keep], s_arr[keep],
                   s=dot_sizes[keep],
                   c=k_colors[keep],
                   alpha=0.7, edgecolors='k', linewidths=0.3)

    legend_patches = [Patch(color=c, label=n) for n, c in window_colors.items()]
    ax.legend(handles=legend_patches, fontsize=7, loc='upper right')

plt.tight_layout()
plt.show()
```

```python
# Correct phasor normalization
g_arr2 = np.zeros(len(k_bins2))
s_arr2 = np.zeros(len(k_bins2))

for i in range(len(k_bins2)):
    ts = DELTA_PHI_k2[i]
    ts_dt = ts - np.mean(ts)
    F = fft(ts_dt)
    # Simple, correct normalization: divide by F[0] of the
    # non-detrended series (total area)
    ts_pos = ts - ts.min()  # shift to positive
    F_pos = fft(ts_pos)
    norm = np.abs(F_pos[0])
    if norm == 0:
        continue
    g_arr2[i] = np.real(F[omega_idx]) / norm
    s_arr2[i] = np.imag(F[omega_idx]) / norm

print(f"g range: {g_arr2.min():.4f} to {g_arr2.max():.4f}")
print(f"s range: {s_arr2.min():.4f} to {s_arr2.max():.4f}")

# Plot
theta = np.linspace(0, np.pi, 300)
plt.figure(figsize=(7,6))
plt.plot(0.5 + 0.5*np.cos(theta), 0.5*np.sin(theta),
         'k--', alpha=0.5, label='Semicircle')
plt.scatter(g_arr2, s_arr2,
            s=dot_sizes,
            c=k_colors,
            alpha=0.7, edgecolors='k', linewidths=0.3)
plt.xlim(-0.2, 1.2)
plt.ylim(-0.2, 0.7)
plt.xlabel('g')
plt.ylabel('s')
plt.title('Phasor - corrected normalization')
plt.axhline(0, color='grey', lw=0.5)
plt.legend()
plt.show()
```
